## 1. Pertes de substitution

### 1.1 Pourquoi ne pas optimiser directement la perte 0-1 ?

En classification binaire, avec $y \in \{-1, +1\}$ et un score $h(x) \in \mathbb{R}$ (par exemple $h(x) = \langle w, x\rangle$ pour un modele lineaire), l'erreur de classification s'ecrit naturellement a l'aide de la **perte 0-1** :

$$ \ell_{0-1}(h(x), y) = \mathbb{1}\{y\,h(x) \leq 0\}. $$

On l'exprime en fonction de la **marge** $y\,h(x)$ : la marge est positive si et seulement si la prediction est correcte (le signe de $h(x)$ correspond a $y$).

Le probleme : la perte 0-1, vue comme fonction de $u = y\,h(x)$, est **non convexe** et **non differentiable** en $u = 0$ (elle vaut $1$ pour $u \le 0$ et $0$ pour $u > 0$). Minimiser directement le risque empirique associe est un probleme combinatoire NP-difficile en general : on ne peut pas utiliser de descente de gradient dessus. On remplace donc la perte 0-1 par une **perte de substitution** (*surrogate loss*), convexe (et parfois differentiable), qui la majore et qui est facile a optimiser.

### 1.2 Les trois pertes de substitution usuelles

- **Hinge loss** (utilisee par le SVM) :
$$ \ell_{\text{hinge}}(h(x), y) = [1 - y\,h(x)]_+ = \max(0,\ 1 - y\,h(x)). $$
  Elle est nulle des que la marge $y\,h(x) \geq 1$ (exemple bien classe avec une marge confortable), et croit lineairement sinon.

- **Perte exponentielle** (utilisee en boosting) :
$$ \ell_{\exp}(h(x), y) = \exp(-y\,h(x)). $$
  Elle donne un poids plus important aux erreurs (croissance exponentielle quand $y\,h(x) \to -\infty$) et **ne s'annule jamais**, meme pour les exemples tres bien classes.

- **Perte logistique** (utilisee par la regression logistique), pour $y \in \{-1, 1\}$ :
$$ \ell_{\text{log}}(h(x), y) = \frac{1}{\ln 2}\, \ln\big(1 + \exp(-y\,h(x))\big). $$
  Le facteur $1/\ln 2$ normalise la perte pour qu'elle vaille exactement $1$ en $y\,h(x) = 0$ (prediction totalement incertaine), ce qui permet de la comparer plus facilement a la perte 0-1 sur le meme graphique.

Les trois pertes ci-dessus sont convexes en $u = y\,h(x)$ et majorent (ou sont tres proches de majorer) la perte 0-1 : les minimiser revient donc, en un certain sens, a controler l'erreur de classification.

```python
def perte_01(u):
    """Perte 0-1 en fonction de la marge u = y * h(x)."""
    return (u <= 0).astype(float)

def perte_hinge(u):
    """Hinge loss (SVM)."""
    return np.maximum(0.0, 1.0 - u)

def perte_exponentielle(u):
    """Perte exponentielle (boosting)."""
    return np.exp(-u)

def perte_logistique(u):
    """Perte logistique (regression logistique), normalisee par 1/ln(2)."""
    return np.log1p(np.exp(-u)) / np.log(2)

u = np.linspace(-3, 3, 400)

fig, ax = plt.subplots()
ax.plot(u, perte_01(u), label="Perte 0-1", color="black", linewidth=2)
ax.plot(u, perte_hinge(u), label="Hinge loss (SVM)", color="tab:blue")
ax.plot(u, perte_logistique(u), label="Perte logistique", color="tab:orange")
ax.plot(u, perte_exponentielle(u), label="Perte exponentielle (boosting)", color="tab:green")
ax.axvline(0, color="gray", linestyle=":", linewidth=1)
ax.set_xlabel("Marge $y \\cdot h(x)$")
ax.set_ylabel("Valeur de la perte")
ax.set_title("Perte 0-1 et pertes de substitution en fonction de la marge")
ax.set_ylim(0, 3)
ax.legend()
ax.text(-2.8, 2.7, "Prediction incorrecte", fontsize=9, color="gray")
ax.text(1.1, 2.7, "Prediction correcte", fontsize=9, color="gray")
plt.show()
```

## 2. k plus proches voisins (k-NN)

### 2.1 Principe de l'algorithme

Le k-NN [Cover et Hart, 1967] est une methode **non parametrique** : elle ne fait aucune hypothese sur la distribution sous-jacente des donnees. Pour predire l'etiquette d'une nouvelle observation $x'$ :

1. on calcule la distance entre $x'$ et chaque exemple $x_i$ de l'ensemble d'apprentissage $S = \{(x_i, y_i)\}_{i=1}^m$ ;
2. on trie ces distances par ordre croissant et on retient les $k$ exemples les plus proches (les $k$ plus proches voisins de $x'$) ;
3. on predit l'etiquette **majoritaire** parmi ces $k$ voisins :
$$ \hat{y}(x') = \arg\max_{y \in \mathcal{Y}} k_y, $$
ou $k_y$ est le nombre de voisins, parmi les $k$ plus proches de $x'$, qui appartiennent a la classe $y$.

**Cas particulier $k=1$** : Cover et Hart (1967) ont montre que, lorsque la taille $m$ de l'echantillon est suffisamment grande, le taux d'erreur du 1-NN est au plus egal a **deux fois l'erreur de Bayes** (l'erreur minimale atteignable compte tenu de la distribution des donnees).

**Effet de k** : plus k est petit, plus la frontiere de decision est sensible aux points individuels (risque de sur-apprentissage / forte variance) ; plus k est grand, plus la decision est lissee (risque de sous-apprentissage / fort biais), jusqu'a $k = m$ ou l'on predit systematiquement la classe majoritaire de tout le jeu de donnees, quelle que soit la requete. En pratique, on choisit k par **validation croisee**.

### 2.2 Distance usuelle

La distance la plus utilisee est la distance euclidienne (norme $L_2$), pour $x, x' \in \mathbb{R}^d$ :
$$ d_2(x, x') = \|x - x'\|_2 = \sqrt{\sum_{j=1}^d (x_j - x'_j)^2}. $$
On peut aussi utiliser la distance de Manhattan (norme $L_1$) $d_1(x, x') = \sum_{j=1}^d |x_j - x'_j|$, ou plus generalement la norme $L_p$ pour $p \geq 1$ :
$$ d_p(x, x') = \Big(\sum_{j=1}^d |x_j - x'_j|^p\Big)^{1/p}. $$

```python
# Un petit exemple 1D pour se familiariser avec le vote majoritaire du k-NN.
X_train_1d = np.array([1, 2, 3, 5, 8]).reshape(-1, 1)
y_train_1d = np.array([1, 1, 0, 0, 1])
x_query = 4

distances = np.abs(X_train_1d.ravel() - x_query)
ordre = np.argsort(distances)
print("Points tries par distance croissante a x' = 4 :")
for idx in ordre:
    print(f"  x={X_train_1d[idx, 0]}, y={y_train_1d[idx]}, distance={distances[idx]}")

for k in [1, 3, 5]:
    voisins = ordre[:k]
    votes = y_train_1d[voisins]
    classe, effectifs = np.unique(votes, return_counts=True)
    prediction = classe[np.argmax(effectifs)]
    print(f"k={k} -> voisins retenus: {sorted(X_train_1d[voisins].ravel())}, "
          f"vote majoritaire = {prediction}")
```

### 2.3 Effet de k sur la frontiere de decision (compromis biais-variance)

Regardons visuellement ce qui se passe sur un jeu de donnees non lineairement separable (`make_moons`, avec un peu de bruit) lorsqu'on fait varier k.

```python
X_moons, y_moons = make_moons(n_samples=300, noise=0.25, random_state=0)

xx, yy = np.meshgrid(
    np.linspace(X_moons[:, 0].min() - 0.5, X_moons[:, 0].max() + 0.5, 300),
    np.linspace(X_moons[:, 1].min() - 0.5, X_moons[:, 1].max() + 0.5, 300),
)

valeurs_k = [1, 5, 15, 50]
fig, axes = plt.subplots(1, len(valeurs_k), figsize=(18, 4.5), sharex=True, sharey=True)

for ax, k in zip(axes, valeurs_k):
    clf = KNeighborsClassifier(n_neighbors=k).fit(X_moons, y_moons)
    Z = clf.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    ax.contourf(xx, yy, Z, alpha=0.3, cmap="coolwarm")
    ax.scatter(X_moons[:, 0], X_moons[:, 1], c=y_moons, cmap="coolwarm",
               edgecolor="k", s=15)
    ax.set_title(f"k = {k}")
    ax.set_xlabel("$x_1$")
axes[0].set_ylabel("$x_2$")
fig.suptitle("Frontieres de decision du k-NN selon k (make_moons) : "
             "sur-apprentissage (k petit) -> sous-apprentissage (k grand)")
plt.tight_layout()
plt.show()
```

On observe qu'avec `k=1`, la frontiere est tres irreguliere : elle colle aux points d'apprentissage, y compris au bruit (**sur-apprentissage**, forte variance). Avec `k=50`, la frontiere devient tres lisse, presque lineaire, et ne capture plus la structure en « croissants » des donnees (**sous-apprentissage**, fort biais). Une valeur intermediaire (ici `k=5` ou `k=15`) offre un meilleur compromis.

### 2.4 Le fleau de la dimension (*curse of dimensionality*)

Le k-NN repose entierement sur la notion de distance : sa pertinence suppose donc que la notion de « proximite » reste informative. Or, en grande dimension, un phenomene contre-intuitif apparait : les distances entre points tires aleatoirement ont tendance a se **concentrer** autour d'une meme valeur, si bien que la difference entre le point le plus proche et le point le plus loin devient negligeable devant la distance elle-meme. La notion de « voisin proche » perd alors de son sens.

On l'illustre en tirant $n$ points uniformement dans l'hypercube $[0, 1]^d$, en calculant la distance euclidienne d'un point de requete fixe (le centre de l'hypercube) a tous les autres points, puis en regardant le ratio :
$$ \frac{d_{\max} - d_{\min}}{d_{\min}} $$
en fonction de la dimension $d$.

```python
rng = np.random.default_rng(0)
n_points = 1000
dimensions = [1, 2, 5, 10, 20, 50, 100, 200, 500, 1000]
ratios = []

for d in dimensions:
    X_cube = rng.uniform(0, 1, size=(n_points, d))
    requete = np.full(d, 0.5)  # centre de l'hypercube
    distances_cube = np.linalg.norm(X_cube - requete, axis=1)
    d_min, d_max = distances_cube.min(), distances_cube.max()
    ratios.append((d_max - d_min) / d_min)

plt.figure()
plt.plot(dimensions, ratios, marker="o")
plt.xscale("log")
plt.xlabel("Dimension d (echelle log)")
plt.ylabel("$(d_{max} - d_{min}) / d_{min}$")
plt.title("Fleau de la dimension : effondrement du contraste des distances")
plt.show()

print("Ratio (d_max - d_min)/d_min pour chaque dimension :")
for d, r in zip(dimensions, ratios):
    print(f"  d={d:>4d} -> ratio={r:.4f}")
```

On observe que ce ratio, tres grand en petite dimension (le point le plus proche et le point le plus loin sont tres differents), s'effondre rapidement quand $d$ augmente : en tres grande dimension, tous les points sont « a peu pres a la meme distance » du point de requete. Cela explique pourquoi le k-NN (et plus generalement toute methode fondee sur les distances) devient peu performant en tres grande dimension, surtout lorsque de nombreuses variables sont peu ou pas informatives.

### 2.5 Mise a l'echelle des variables

Un probleme lie a la notion de distance : si les descripteurs sont exprimes dans des unites ou des echelles tres differentes, la distance euclidienne peut etre **dominee** par le descripteur ayant la plus grande echelle, meme s'il n'est pas le plus informatif.

**Exemple.** Avec $x^{(1)} = (0.3,\ 0.89,\ 232)$ et $x^{(2)} = (-0.1,\ 0.23,\ 652)$ :
$$ d_2^2(x^{(1)}, x^{(2)}) = (0.3+0.1)^2 + (0.89-0.23)^2 + (232-652)^2 \approx (232-652)^2, $$
les deux premiers descripteurs n'ont quasiment aucun impact sur la distance : tout est domine par le troisieme.

On corrige generalement ce probleme en mettant les variables a la meme echelle, par exemple par **normalisation** (centrage-reduction) :
$$ v_j \leftarrow \frac{v_j - \mu_j}{\sigma_j}, $$
ou $\mu_j$ et $\sigma_j$ sont la moyenne et l'ecart-type du descripteur $j$ (c'est ce que fait `StandardScaler` de `scikit-learn`), ou par une mise a l'echelle min-max $v_j \leftarrow \dfrac{v_j - \min v_j}{\max v_j - \min v_j}$ (intervalle $[0, 1]$).

Illustrons l'effet de cette mise a l'echelle sur les performances du k-NN, en ajoutant artificiellement a `load_wine` un descripteur bruite sur une echelle tres differente des autres.

```python
wine = load_wine()
X_wine, y_wine = wine.data, wine.target

# Les variables de `load_wine` ont deja des echelles tres heterogenes
# (ex: 'proline' varie dans les centaines, 'hue' varie entre 0 et ~1.7).
print("Min/Max par variable (extrait) :")
for name, col in list(zip(wine.feature_names, X_wine.T))[:3] + [
        (wine.feature_names[-1], X_wine[:, -1])]:
    print(f"  {name:25s}: min={col.min():8.2f}, max={col.max():8.2f}")

X_train_w, X_test_w, y_train_w, y_test_w = train_test_split(
    X_wine, y_wine, test_size=0.3, random_state=0, stratify=y_wine)

# Sans mise a l'echelle
clf_brut = KNeighborsClassifier(n_neighbors=5).fit(X_train_w, y_train_w)
acc_brut = accuracy_score(y_test_w, clf_brut.predict(X_test_w))

# Avec mise a l'echelle (StandardScaler ajuste seulement sur le train)
scaler = StandardScaler().fit(X_train_w)
X_train_w_sc = scaler.transform(X_train_w)
X_test_w_sc = scaler.transform(X_test_w)
clf_scale = KNeighborsClassifier(n_neighbors=5).fit(X_train_w_sc, y_train_w)
acc_scale = accuracy_score(y_test_w, clf_scale.predict(X_test_w_sc))

print(f"\nExactitude k-NN (k=5) SANS mise a l'echelle : {acc_brut:.3f}")
print(f"Exactitude k-NN (k=5) AVEC StandardScaler    : {acc_scale:.3f}")

plt.figure()
plt.bar(["Sans mise a l'echelle", "Avec StandardScaler"], [acc_brut, acc_scale],
        color=["tab:red", "tab:blue"])
plt.ylabel("Exactitude sur le jeu de test")
plt.title("Effet de la mise a l'echelle des variables sur le k-NN (load_wine)")
plt.ylim(0, 1)
plt.show()
```

On constate que la mise a l'echelle ameliore nettement l'exactitude du k-NN sur ce jeu de donnees : sans elle, les variables a forte amplitude (comme `proline`) dominent le calcul de la distance et masquent l'information portee par les variables a faible amplitude.

### 2.6 Cout algorithmique et raffinements

Pour predire l'etiquette d'une nouvelle observation, le k-NN doit calculer la distance a **tous** les exemples d'apprentissage et garder **tout** le jeu d'apprentissage en memoire : pour $n$ exemples de dimension $d$, la complexite en temps est en $O(nd + nk)$ (calcul des distances, puis selection des $k$ plus proches) et la memoire necessaire est en $O(n)$. Cela rend le k-NN « brut » couteux sur de tres grands jeux de donnees. Des raffinements existent pour reduire ce cout : structures d'indexation spatiale (arbres kd, *ball trees*...), inegalite triangulaire, ou selection d'un sous-ensemble representatif de l'ensemble d'apprentissage (*Condensed Nearest Neighbor*, Hart 1968).

### 2.7 k-NN pondere

Dans le vote majoritaire « classique », chaque voisin compte pour une voix, quelle que soit sa distance a la requete. Le **k-NN pondere** [Dudani, 1976] attribue a chaque voisin un poids, generalement l'**inverse de sa distance** ($1/d_i$), de sorte que les voisins les plus proches pesent davantage dans le vote que les voisins plus eloignes. C'est l'option `weights='distance'` de `KNeighborsClassifier` (`weights='uniform'` par defaut correspond au vote non pondere).

**Exemple.** Pour classer $x'$ avec le 3-NN, avec des voisins d'etiquettes $y_1=1, y_2=1, y_3=-1$ situes a des distances $d_1=3, d_2=4, d_3=1$ : le vote majoritaire non pondere donne 2 voix a la classe $1$ contre 1 voix a la classe $-1$, donc predit $1$. Avec un poids $1/d_i$, la classe $1$ recoit $1/3 + 1/4 = 7/12 \approx 0.58$ et la classe $-1$ recoit $1/1 = 1$ : c'est la classe $-1$ qui l'emporte.
